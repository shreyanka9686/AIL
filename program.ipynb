```python
import pandas as pd 
data={
    'name':['anusha','shreya','anu','divya','thanu'],
    'age':[18,19,20,17,21],
    'branch':['cs','ec','civil','cp','amt'], 
    'science':[86,45,20,95,50],
    'maths':[55,68,15,80,65], 
    'english':[88,80,24,75,89]
}
student=pd.DataFrame(data)
student['total']=student[['science','maths','english']].sum(axis=1)
student['average']=student[['science','maths','english']].mean(axis=1)
student['reuslt']=student['average'].apply(lambda avg:'pass' if avg>=50 else 'fail')
student.to_csv('student_results.csv',index=False)
print(student)
```

```python
import numpy as np
tm=np.random.uniform(low=20.0, high=40.0, size=50)
avg_tm=np.mean(tm)
max_tm=np.max(tm)
min_tm=np.min(tm)
nrml_tm=(tm-min_tm)/(max_tm-min_tm)
frnt_tm=list(map(lambda c: (c*9/5)+32,tm))
high_tm=list(filter(lambda t: t>35.0,tm))
print("temperature:")
print(tm)
print(f"avg tm:{avg_tm:.2f}")
print(f"max tm:{max_tm:.2f}")
print(f"min tm:{min_tm:.2f}\n")
print("first 10 normalized readings:")
print(nrml_tm[:10])
print("\n first 5 fahrenheit readings is:",frnt_tm[:5])
print("\n temperatures >35:",high_tm)
```

```python
import pandas as pd 
data={
    'name':['sneha','sahana','kavan','bhanu'],
    'maths':[69,45,67,87],
    'science':[79,80,90,76], 
    'english':[67,87,67,80]
}
df=pd.DataFrame(data)
df.to_csv('student_scores.csv',index=False)
df_read=pd.read_csv('student_scores.csv')
df['total']=df[['maths','science','english']].sum(axis=1)
df['average']=df[['maths','science','english']].mean(axis=1)
print("First 3 row:\n",df_read.head(3))
print("\ncolumns:",df_read.columns.tolist())
print("\nshape of DataFrame:",df_read.shape)
print(df)
records=df.to_dict(orient='records')
result=list(filter(lambda student:student['science']>=80,records))
science=[student['science']for student in result]
print("science marks generate than 80:",science)
sorted=df.sort_values(by='english',ascending=False)
print(sorted)
df.to_csv("data.csv",index=False)
```
