```python
Q1=df['Rent'].quantile(0.25)
Q3=df['Rent'].quantile(0.75)
iqr=Q3-Q1 
lower_bound=Q1-1.5*iqr 
upper_bound=Q3+1.5*iqr
df_no_outliers=df[(df['Rent'] >= lower_bound) & (df['Rent'] <= upper_bound)]
df_no_outliers['Rent'].plot(kind='box')
```
