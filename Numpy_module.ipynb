## 1.Using numpy module perform following operations:
### a)Demonstrate array Aggregative functions:

```python
import numpy as np
arr = np.array([1,2,3,4,5])
print("array elements are:",arr)
print("mimimum value is:",arr.min())
print("maximum value is:",arr.max())
print("sum of the number is:",arr.sum())
print("average of the number is:",arr.mean())
```

### b) Demonstrate Vectorized operations:

```python
import numpy as np
arr=np.array([10,20,30,40,50])
result=arr*2
print("the multiple by the 2:",result)
result1=arr**2
print("the square of the 2 numbers is:",result1)
result2=arr-2
print("the substraction of the 2 number is:",result2)
```

### Lambda

```python
result=lambda a,b:a+b
print("the addition is:",result(5,6))
result1=lambda a,b:a-b
print("the substraction is:",result1(9,2))
result2=lambda a,b:a*b
print("the multiplication is:",result2(5,6))
result3=lambda a,b:a/b
print("the divition is:",result3(6,2))
result4=lambda a:a**2
print("the square is:",result4(12))
```

## Filter

#### Select items based on a condition:

```python
def even (x):
    if x%2==0:
      return True
L1=[2,3,4,5,6,7,8]
even=list(filter(even,L1))
even
```

```python
def lessthan(a):
    if a<=5000:
        return True
salary=[5000,7000,2000,8000,3000]
lessthan=list(filter(lessthan,salary))
lessthan
```

## Reduce

#### Combine items into a single value

```python
from functools import reduce
def add(x,y):
  return x+y
L1=[12,13,14,15,16]
result=reduce(add,L1)
result
```

```python
from functools import reduce
def mean(x,y):
    return x+y
L=[12,13,14,15,16,23]
result=reduce(mean,L)
print(result/6)
```

```python
from functools import reduce
def mult(x,y):
    return x*y
list=[2,3,4,5]
result=reduce(mult,list)
result
```
