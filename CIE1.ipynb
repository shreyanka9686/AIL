### *Plot the count of each species on bar graph.

```python
count=df['Species'].value_counts()
count
```

```python
count.plot(kind='bar')
plt.ylabel("count of the values")
plt.title("Bar graph distribution")
plt.show()
```

### *plot the histogram of petal length.

```python
counts=df['PetalLengthCm'].value_counts()
counts
```

```python
counts.plot(kind='hist')
plt.xlabel("PetalLengthCm")
plt.title("Histogram distribution")
plt.show()
```

### *Plot the scatter plot for sepal length and sepal width with different colours for each species.

```python
cou=df[['SepalLengthCm','SepalWidthCm']].value_counts()
cou
```

```python
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'])
plt.xlabel('SepalLengthCm SepalWidthCm')
plt.ylabel("value counts")
plt.title("Scatter plot distribution")
plt.show()
```
