## 2.Using pandas module perform the following operations:

```python
import pandas as pd
import numpy as np
marks = {'Chemistry': [67,90,66,32],
'Physics': [45,92,72,40],
'Mathematics': [50,87,81,12],
'English': [19,90,72,68]}
marks_df = pd.DataFrame(marks, index = ['Subodh', 'Ram', 'Abdul', 'John'])
marks_df
```

### a) Aggregation and grouping,time series operations and pivot and melt functions:

### Aggreagations functions:

```python
marks_df['Chemistry'].agg(['mean','median','max','min','std','sum'])
```

### Grouping

```python
marks_df.groupby('Physics')['Mathematics'].sum()
```

```python
a=marks_df.groupby(['Chemistry','English']).sum()
a
```

### Time series

```python
range_date=pd.date_range(start='1/1/2020',end='1/10/2025',freq='D')
range_date
```

```python
month=pd.date_range(start='1/1/2020',end='1/10/2025',freq='ME')
month
```

### b) Demonstrate map,filter,reduce and lambda functions with dataframe

```python
Grade_marks=list(map(lambda x:x+10,marks_df['Chemistry']))
print("Add Grade marks to Chemistry subject:")
Grade_marks
```
