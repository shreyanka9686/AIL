```python
import pandas as pd 
data={
    'Ename':(["raj","ram","rahim","roja","abhi"]), 
    'Type':(["regular","adoc","contract","regular","contract"]), 
    'Department':(["cs","ec","ec","cs","cp"]), 
    'Expiriance':([2,8,5,10,5]), 
    'Salary':([10000,20000,5000,15000,6000])
}
df=pd.DataFrame(data)
df
```

```python
agg=df.pivot_table(values='Salary',index='Type',aggfunc=('sum','mean','count'))
agg
```

```python
avg_sal=df.pivot_table(values='Salary',index='Department',columns='Type',aggfunc='mean')
avg_sal
```
