# Consider Titanic dataset perform univariate , bivariate & plot the graph

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('C:/Users/ADMIN/Downloads/titanic.csv')
df.head()
```

```python
df['Pclass'].value_counts().plot(kind='bar')
plt.xlabel("Pclass")
plt.ylabel("count")
plt.title("Gender distribution")
```

```python
a=df[['Pclass','Sex']].value_counts().unstack()
a.plot(kind='bar')
```
