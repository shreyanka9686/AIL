### Create a dataset and perform the following.
#### Univariate Analysis Practice (Matplotlib).
### 1.Plot a histogram of Marks_Math
#### use plt.hist()¶

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
data={
    'ID':[1,2,3,4,5,6,7,8], 
    'Name':['Riya','Arjun','Sneha','Rakesh','Aditi','Kiran','Anjali','Manish'], 
    'Gender':['F','M','F','M','F','M','F','M'],
    'Age':[17,18,17,19,18,18,17,19], 
    'Study_Hours':[3.5,2.0,4.0,1.5,3.8,2.5,4.2,1.0],
    'Attedence(%)':[95,82,98,70,92,85,99,65],
    'Marks_math':[88,75,90,65,95,78,91,60], 
    'Marks_Science':[92,80,89,68,93,75,95,58], 
    'Grade':['A','B','A','C','A','B','A','D']
   }
df=pd.DataFrame(data) 
df
```

```python
plt.hist(df['Marks_math']) 
plt.title("Histogram of Marks in Math")
plt.xlabel("Marks_math")
plt.ylabel("Frequency")
plt.show()
```

### 2.Draw a bar chart of number of students per grade.

```python
value_count=df['Grade'].value_counts()
plt.bar(value_count.index,value_count.values) 
plt.title("Bar chart")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.show()
```

### 3.Show the distribution of Study_Hours.

```python
plt.hist(df['Study_Hours'])
plt.title("Histogram for Study Hours:")
plt.xlabel("Student_Hours")
plt.ylabel("Students")
plt.show()
```

```python
plt.boxplot(df['Study_Hours'])
plt.show()
```

### 4.plot a pie chart of gender distribution.

```python
gender=df['Gender'].value_counts()
plt.pie(gender.values,labels=gender.index)
plt.title("Gender Distribution:")
plt.show()
```

### 5.plot the frequency of attedence percentage

```python
plt.hist(df['Attedence(%)'],bins=5,edgecolor='black')
plt.title("Frequency of Attedence percentage:")
plt.xlabel("Attedence Percentage:")
plt.ylabel("Frequency")
plt.show()
```

### Bivariate Analysis Practice
#### (Matplotlib)

### 6.scatter plot between Study_Hours and Marks_Math.

```python
plt.scatter(df['Study_Hours'],df['Marks_math'])
plt.title("scatter plot betwee Hours and Marks in Math")
plt.xlabel("Study Hours")
plt.ylabel("Marks in Math")
plt.show()
```

### 7.Scatter plot between Attedence(%)and Marks_Science.

```python
plt.scatter(df['Attedence(%)'],df['Marks_Science'])
plt.title("scatter plot betwee Attedence(%) and Marks in Science")
plt.xlabel("Attedence(%)")
plt.ylabel("Marks in Science")
plt.show()
```

### 8.Compare Marks_Math by Gender using side-by-side bar char.

```python
gen=df.groupby('Gender')['Marks_math'].mean()
plt.bar(gen.index,gen.values)
plt.title("Average Marks in Math by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks in Math")
plt.show()
```

### 9.Line Plot showing marks of each student in math science.

```python
plt.plot(df['Name'],df['Marks_math'],label='math')
plt.plot(df['Name'],df['Marks_Science'],label='Science')
plt.title("Line plot showing Marks math and Marks in Science")
plt.xlabel("Name")
plt.ylabel("Math and Science")
plt.legend()
plt.show()
```

### 10.Boxplot  comparing Marks_math across Grade.

```python
plt.boxplot([df[df['Grade']=='A']['Marks_math'],df[df['Grade']=='B']['Marks_math'],df[df['Grade']=='C']['Marks_math']],labels=['A','B','C'])
plt.show()
```

```python
plt.plot(df['Name'],df[['Marks_math','Marks_Science']],label='math')
plt.title("Line plot showing Marks math and Marks in Science")
plt.xlabel("Name")
plt.ylabel("Math and Science")
plt.legend()
plt.show()
```
