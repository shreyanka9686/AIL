## Consider mall customer dataset & perform the following operations

```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("C:/Users/ADMIN/Downloads/Mall_Customers.csv")
df
```

```python
df.isnull().sum()
```

```python
x=df[['Annual Income (k$)','Spending Score (1-100)']].values
x
```

```python
x=StandardScaler().fit_transform(x)
```

```python
x
```

### A) Using Elbow method find the optimal number of cluster

```python
wcss=[]
for k in range(1,11):
    km=KMeans(n_clusters=k,random_state=0)
    km.fit(x)
    wcss.append(km.inertia_)
plt.plot(range(1,11),wcss,marker='o')  
plt.title('elbow')
plt.xlabel('k')
plt.ylabel('wcss')
plt.show()
```

### b).Train K-means cluster algorithom & training dataset

```python
kmeans=KMeans(n_clusters=5,random_state=0)
labels=kmeans.fit_predict(x)

labels
```

```python
plt.scatter(x[:,0],x[:,1],c=labels)
plt.title('cluster') 
plt.show()

```

### C).Evaluate the model using inertia & other matrix (silhouette_score)

```python
print("The inertia for 5 cluster is",kmeans.inertia_)

```

```python
from sklearn.metrics import silhouette_score
score=silhouette_score(x,labels)
score
```
