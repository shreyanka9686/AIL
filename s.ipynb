```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv("C:/Users/ADMIN/Downloads/titanic (1).csv")
```

```python
df.head()
```

```python
df['Age'].plot(kind='box')
```

```python
Q1=df['Age'].quantile(0.25)
Q3=df['Age'].quantile(0.75)
iqr=Q3-Q1 
ub=Q3+(1.5*iqr)
lb=Q1-(1.5*iqr)
uc=65
lc=0
for i in df['Age']: 
    if i < lb:
        df['Age']=df['Age'].replace(i,lc)
    elif i>ub:
        df['Age']=df['Age'].replace(i,uc) 
df['Age'].plot(kind='box')
```
