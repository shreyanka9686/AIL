# Consider titanic dataset & plot different graph to visualization, trends, propotions, categorical data,uncertainities & distributions

```python
#import libraries
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sb 
```

```python
df= pd.read_csv('C:/Users/ADMIN/Downloads/titanic.csv')
df.head()
```

```python
df['Age'].plot(kind='box')
```

```python
df['Pclass'].value_counts().plot(kind='bar')
plt.xlabel("Pclass")
plt.ylabel("count")
plt.title("Pclass distribution")
```

```python
plt.scatter(x='Ticket',y='Age',data=df)
plt.xlabel("number of Ticket")
plt.ylabel("Age")
plt.title("scatter plot distributed ")
plt.show()
```

```python
plt.hist(df['Age'])
plt.xlabel('Age')
plt.ylabel('counts')
plt.title('histogream distribution')
```

```python
data=(df['Embarked'].value_counts())
data.plot(kind='pie',autopct='%1.2f%%')
```

```python
plt.plot(df['PassengerId'])
plt.show()
```
